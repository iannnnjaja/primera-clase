```python
import json
import os

ARCHIVO = "cuentas.json"

# CARGAR DATOS
def cargar_datos():
    if os.path.exists(ARCHIVO):
        try:
            with open(ARCHIVO, "r") as archivo:
                return json.load(archivo)
        except (json.JSONDecodeError, ValueError):
            return []
    return []

# GUARDAR DATOS
def guardar_datos(datos):
    with open(ARCHIVO, "w") as archivo:
        json.dump(datos, archivo, indent=4)

# ALTAS
def altas(datos):
    print("\n--- ALTA DE CLIENTE ---")
    try:
        nombre = input("Nombre: ")
        edad = int(input("Edad: "))
        no_cuenta = int(input("Número de cuenta: "))
        monto = float(input("Monto inicial: "))

        cliente = {
            "nombre": nombre,
            "edad": edad,
            "no_cuenta": no_cuenta,
            "monto": monto
        }

        datos.append(cliente)
        guardar_datos(datos)
        print("Cliente agregado correctamente.\n")
    except ValueError:
        print("Error: Ingrese valores numéricos válidos para edad, cuenta y monto.\n")

# BAJAS
def bajas(datos):
    print("\n--- BAJA DE CLIENTE ---")
    try:
        cuenta = int(input("Número de cuenta a eliminar: "))
        encontrado = False

        for cliente in datos:
            if cliente["no_cuenta"] == cuenta:
                datos.remove(cliente)
                guardar_datos(datos)
                encontrado = True
                print("Cliente eliminado correctamente.\n")
                break

        if not encontrado:
            print("Cuenta no encontrada.\n")
    except ValueError:
        print("Error: Ingrese un número de cuenta válido.\n")

# MOSTRAR CLIENTES
def mostrar_clientes(datos):
    print("\n--- CLIENTES REGISTRADOS ---")
    if len(datos) == 0:
        print("No hay clientes.\n")
        return

    for cliente in datos:
        print(f"""
        Nombre: {cliente['nombre']}
        Edad: {cliente['edad']}
        No. Cuenta: {cliente['no_cuenta']}
        Monto: ${cliente['monto']:.2f}
        --------------------------
        """)

# INTERÉS COMPUESTO
def interes_compuesto(datos):
    print("\n--- INTERÉS COMPUESTO ---")
    try:
        cuenta = int(input("Número de cuenta: "))
        encontrado = False

        for cliente in datos:
            if cliente["no_cuenta"] == cuenta:
                P = cliente["monto"]
                meses = int(input("¿Cuántos meses? "))
                tasa = float(input("Tasa anual (%): "))

                r = tasa / 100
                n = 12
                t = meses / 12

                A = P * (1 + r / n) ** (n * t)

                print(f"""
                Cliente: {cliente['nombre']}
                Capital inicial: ${P:.2f}
                Monto final: ${A:.2f}
                Ganancia: ${A - P:.2f}
                """)
                encontrado = True
                break

        if not encontrado:
            print("Cuenta no encontrada.\n")
    except ValueError:
        print("Error: Ingrese datos numéricos válidos.\n")

# MENÚ PRINCIPAL
def menu():
    datos = cargar_datos()

    while True:
        print("""
        ========= MENÚ =========
        1. Altas
        2. Bajas
        3. Mostrar clientes
        4. Calcular interés compuesto
        5. Salir
        ========================
        """)
        opcion = input("Selecciona una opción: ")

        if opcion == "1":
            altas(datos)
        elif opcion == "2":
            bajas(datos)
        elif opcion == "3":
            mostrar_clientes(datos)
        elif opcion == "4":
            interes_compuesto(datos)
        elif opcion == "5":
            print("Programa finalizado.")
            break
        else:
            print("Opción inválida.\n")

# INICIO
if __name__ == "__main__":
    menu()
```
