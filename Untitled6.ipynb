```python
class Persona:

    def __init__(self, nombre="", edad=0, direccion=""):
        self.nombre = nombre
        self.edad = edad
        self.direccion = direccion


    def imprime(self):
        print("Nombre:", self.nombre)
        print("Edad:", self.edad)
        print("Dirección:", self.direccion)


class Cliente(Persona):

    def __init__(self, nombre="", edad=0, direccion="", numCliente=0, cel=0, servicio=""):
        super().__init__(nombre, edad, direccion)
        self.numCliente = numCliente
        self.cel = cel
        self.servicio = servicio


    def imprime(self):
        super().imprime()
        print("NumCliente:", self.numCliente)
        print("Cel:", self.cel)
        print("Servicio:", self.servicio)


print("--- Registro de Cliente ---")
nombre = input("Ingresa el nombre: ")
edad = int(input("Ingresa la edad: "))
direccion = input("Ingresa la dirección: ")
numCliente = input("Ingresa el número de cliente: ")
cel = int(input("Ingresa el celular: "))
servicio = input("Ingresa el servicio: ")


cliente1 = Cliente(nombre, edad, direccion, numCliente, cel, servicio)


print("\nDATOS DEL CLIENTE")
cliente1.imprime()
```
